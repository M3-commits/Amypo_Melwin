```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create a new dataset with 10 variables
data = {
    "Attendance": [60, 90, 75, 95, 50, 88, 80, 98, 85, 70],
    "Study_Consistency": [55, 85, 65, 92, 45, 80, 72, 96, 88, 68],
    "Previous_Score": [45, 70, 55, 82, 40, 78, 65, 90, 72, 58],
    "Assignment_Score": [50, 80, 60, 90, 42, 85, 70, 95, 88, 65],
    "Quiz_Score": [48, 75, 58, 85, 38, 80, 68, 92, 82, 62],
    "Midterm_Score": [42, 72, 56, 88, 35, 78, 66, 94, 80, 60],
    "Practical_Score": [55, 82, 68, 91, 45, 86, 74, 97, 89, 70],
    "Project_Score": [60, 85, 70, 93, 50, 88, 76, 96, 90, 72],
    "Class_Participation": [50, 80, 65, 90, 40, 85, 72, 95, 88, 60],
    "Final_Score": [42, 76, 58, 88, 35, 82, 68, 94, 79, 60]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display first five rows
print(df.head())

# Display dataset shape
print("Shape:", df.shape)

# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Create Box Plot
plt.figure(figsize=(12, 6))
sns.boxplot(data=df)

plt.title("Student Performance - Box Plot")
plt.xlabel("Performance Variables")
plt.ylabel("Scores")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
