```python

# SMART STUDENT PERFORMANCE PREDICTION

# STEP 1: Import libraries

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

print("Machine Learning Environment Ready")


# STEP 2: Create dataset

data = {
    "Study_Hours": [2, 3, 5, 1, 6, 4, 2, 7, 3, 8, 5, 1],
    "Attendance": [60, 65, 75, 40, 85, 70, 50, 90, 55, 95, 80, 35],
    "Previous_Marks": [45, 50, 60, 30, 70, 55, 35, 80, 40, 85, 65, 25],
    "Final_Marks": [48, 52, 65, 35, 78, 60, 40, 88, 45, 92, 72, 30],
    "Result": ["Fail", "Pass", "Pass", "Fail", "Pass", "Pass",
               "Fail", "Pass", "Fail", "Pass", "Pass", "Fail"]
}

df = pd.DataFrame(data)

print("\nDataset:")
print(df)


# STEP 3: Select input features

X = df[["Study_Hours", "Attendance", "Previous_Marks"]]


# =================================
# MODEL 1: REGRESSION
# =================================

print("\nTraining Regression Model...")

y_reg = df["Final_Marks"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y_reg, test_size=0.25, random_state=42
)

reg_model = LinearRegression()

reg_model.fit(X_train, y_train)

print("Regression Model Trained Successfully!")


# =================================
# MODEL 2: CLASSIFICATION
# =================================

print("\nTraining Classification Model...")

y_class = df["Result"].map({
    "Fail": 0,
    "Pass": 1
})

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X, y_class, test_size=0.25,
    random_state=42, stratify=y_class
)

class_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000)
)

class_model.fit(X_train_c, y_train_c)

print("Classification Model Trained Successfully!")


# =================================
# STEP 4: GET USER INPUT
# =================================

print("\n================================")
print(" ENTER STUDENT DETAILS")
print("================================")

study_hours = float(input("Enter Study Hours: "))

attendance = float(input("Enter Attendance (%): "))

previous_marks = float(input("Enter Previous Marks: "))


# Create new student data

new_student = pd.DataFrame({
    "Study_Hours": [study_hours],
    "Attendance": [attendance],
    "Previous_Marks": [previous_marks]
})


# =================================
# STEP 5: REGRESSION PREDICTION
# =================================

predicted_marks = reg_model.predict(new_student)

print("\n================================")
print(" REGRESSION PREDICTION")
print("================================")

print("Predicted Final Marks:",
      round(predicted_marks[0], 2))


# =================================
# STEP 6: CLASSIFICATION PREDICTION
# =================================

predicted_result = class_model.predict(new_student)

print("\n================================")
print(" CLASSIFICATION PREDICTION")
print("================================")

if predicted_result[0] == 1:
    print("Student Result: PASS")
else:
    print("Student Result: FAIL")


print("\n================================")
print(" PREDICTION COMPLETED")
print("================================")
```

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    confusion_matrix
)

# ==========================================
# STEP 1: CREATE THE STUDENT DATASET
# ==========================================

data = {
    'Study Hours': [2, 3, 5, 1, 6, 4, 2, 7, 3, 8, 5, 1],
    'Attendance': [60, 65, 75, 40, 85, 70, 50, 90, 55, 95, 80, 35],
    'Previous Marks': [45, 50, 60, 30, 70, 55, 35, 80, 40, 85, 65, 25],
    'Final Marks': [48, 52, 65, 35, 78, 60, 40, 88, 45, 92, 72, 30],
    'Result': ['Fail', 'Pass', 'Pass', 'Fail', 'Pass', 'Pass',
               'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)

# ==========================================
# STEP 2: DATA PREPROCESSING AND INSPECTION
# ==========================================

print("STUDENT DATASET")
print(df)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset Description:")
print(df.describe())

# ==========================================
# STEP 3: SELECT INPUT FEATURES (X)
# ==========================================

X = df[['Study Hours', 'Attendance', 'Previous Marks']]

# ==========================================
# STEP 4: SELECT TARGET VARIABLES
# ==========================================

# Regression target
y_marks = df['Final Marks']

# Classification target
y_result = df['Result'].map({'Fail': 0, 'Pass': 1})

# ==========================================
# STEP 5: SPLIT TRAINING AND TESTING DATA
# ==========================================

X_train, X_test, y_train_marks, y_test_marks, y_train_result, y_test_result = train_test_split(
    X, y_marks, y_result,
    test_size=0.25,
    random_state=42,
    stratify=y_result
)

print("\nTraining Data:", X_train.shape)
print("Testing Data:", X_test.shape)

# ==========================================
# TASK 1: LINEAR REGRESSION
# ==========================================

print("\n========== TASK 1: REGRESSION ==========")

# STEP 6: TRAIN LINEAR REGRESSION MODEL

linear_model = LinearRegression()

linear_model.fit(X_train, y_train_marks)

print("Linear Regression Model Trained Successfully!")

# Test predictions
reg_predictions = linear_model.predict(X_test)

# ==========================================
# TASK 2: LOGISTIC REGRESSION
# ==========================================

print("\n========== TASK 2: CLASSIFICATION ==========")

# STEP 7: TRAIN LOGISTIC REGRESSION MODEL

logistic_model = LogisticRegression(max_iter=1000)

logistic_model.fit(X_train, y_train_result)

print("Logistic Regression Model Trained Successfully!")

# Test predictions
class_predictions = logistic_model.predict(X_test)

# Convert numerical labels into Pass/Fail
class_labels = ['Fail' if x == 0 else 'Pass'
                for x in class_predictions]

print("\nClassification Predictions:")
print(class_labels)

# ==========================================
# STEP 8: ENTER NEW STUDENT DETAILS
# ==========================================

print("\n========== NEW STUDENT DETAILS ==========")

study_hours = float(input("Enter Study Hours: "))

attendance = float(input("Enter Attendance: "))

previous_marks = float(input("Enter Previous Marks: "))

new_student = pd.DataFrame({
    'Study Hours': [study_hours],
    'Attendance': [attendance],
    'Previous Marks': [previous_marks]
})

# ==========================================
# STEP 9: PREDICT FINAL MARKS
# ==========================================

predicted_marks = linear_model.predict(new_student)

print("\n========== REGRESSION RESULT ==========")

print("Predicted Final Marks:",
      round(predicted_marks[0], 2))

# ==========================================
# STEP 10: PREDICT PASS OR FAIL
# ==========================================

predicted_result = logistic_model.predict(new_student)

result = 'Pass' if predicted_result[0] == 1 else 'Fail'

print("\n========== CLASSIFICATION RESULT ==========")

print("Predicted Student Result:", result)

# ==========================================
# STEP 11: EVALUATE BOTH MODELS
# ==========================================

print("\n========== MODEL EVALUATION ==========")

# Regression evaluation
mae = mean_absolute_error(y_test_marks, reg_predictions)

mse = mean_squared_error(y_test_marks, reg_predictions)

r2 = r2_score(y_test_marks, reg_predictions)

print("\nRegression Metrics:")
print("Mean Absolute Error:", round(mae, 2))
print("Mean Squared Error:", round(mse, 2))
print("R2 Score:", round(r2, 2))

# Classification evaluation
accuracy = accuracy_score(y_test_result, class_predictions)

cm = confusion_matrix(y_test_result, class_predictions)

print("\nClassification Metrics:")
print("Accuracy:", round(accuracy * 100, 2), "%")
print("Confusion Matrix:")
print(cm)

# ==========================================
# STEP 12: COMPARE BOTH MODELS
# ==========================================

comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Logistic Regression'],
    'Target Variable': ['Final Marks', 'Pass/Fail'],
    'Prediction Type': ['Numerical', 'Categorical'],
    'Evaluation Metric': ['MAE, MSE, R2', 'Accuracy'],
    'Score': [
        round(mae, 2),
        round(accuracy * 100, 2)
    ]
})

print("\n========== MODEL COMPARISON ==========")
print(comparison)

print("\nPROJECT COMPLETED SUCCESSFULLY!")
```

```python

# SMART STUDENT PERFORMANCE PREDICTION SYSTEM

# STEP 1: Import libraries

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

print("Machine Learning Environment Ready")


# STEP 2: Create dataset

data = {
    "Study_Hours": [2, 3, 5, 1, 6, 4, 2, 7, 3, 8, 5, 1],

    "Attendance": [60, 65, 75, 40, 85, 70, 50, 90, 55, 95, 80, 35],

    "Previous_Marks": [45, 50, 60, 30, 70, 55, 35, 80, 40, 85, 65, 25],

    "Final_Marks": [48, 52, 65, 35, 78, 60, 40, 88, 45, 92, 72, 30],

    "Result": [
        "Fail", "Pass", "Pass", "Fail",
        "Pass", "Pass", "Fail", "Pass",
        "Fail", "Pass", "Pass", "Fail"
    ]
}

df = pd.DataFrame(data)

print("\nStudent Dataset:")
print(df)


# STEP 3: Select input features

X = df[["Study_Hours", "Attendance", "Previous_Marks"]]


# =================================
# MODEL 1: REGRESSION
# =================================

print("\nTraining Regression Model...")

y_reg = df["Final_Marks"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y_reg, test_size=0.25, random_state=42
)

reg_model = LinearRegression()

reg_model.fit(X_train, y_train)

print("Regression Model Trained!")


# =================================
# MODEL 2: CLASSIFICATION
# =================================

print("\nTraining Classification Model...")

y_class = df["Result"]

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X, y_class, test_size=0.25,
    random_state=42, stratify=y_class
)

class_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000)
)

class_model.fit(X_train_c, y_train_c)

print("Classification Model Trained!")


# =================================
# STEP 4: GET USER INPUT
# =================================

print("\n===== ENTER STUDENT DETAILS =====")

study_hours = float(input("Enter Study Hours: "))

attendance = float(input("Enter Attendance: "))

previous_marks = float(input("Enter Previous Marks: "))


# Create new student data

new_student = pd.DataFrame({
    "Study_Hours": [study_hours],
    "Attendance": [attendance],
    "Previous_Marks": [previous_marks]
})


# =================================
# STEP 5: PREDICT OUTPUT
# =================================

predicted_marks = reg_model.predict(new_student)

predicted_result = class_model.predict(new_student)


# =================================
# STEP 6: DISPLAY OUTPUT
# =================================

print("\n===== STUDENT PERFORMANCE PREDICTION =====")

print("Study Hours:", study_hours)

print("Attendance:", attendance)

print("Previous Marks:", previous_marks)

print("\nPredicted Final Marks:",
      round(predicted_marks[0], 2))

print("Predicted Student Result:",
      predicted_result[0])

print("\n===== PREDICTION COMPLETED =====")
```
